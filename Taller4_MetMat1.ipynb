Inciso a.

```python
# Variables
x, y = sp.symbols('x y')

# Coeficientes
a0, a1, a2 = sp.symbols('a0 a1 a2')
b0, b1, b2 = sp.symbols('b0 b1 b2')

# Bases de P2(x) y G2(y)
Bx = [1, x, x**2]
By = [1, y, y**2]

# Polinomios genéricos
p = a0 + a1*x + a2*x**2
q = b0 + b1*y + b2*y**2

# Producto tensorial
prod_tensor = sp.expand(p*q)

# Base tensorial
Base_tensor = [bx*by for bx in Bx for by in By]

# Encontrar los coeficientes c^{ij}
coeff_matrix = sp.Matrix(3, 3, lambda i, j: sp.expand(prod_tensor).coeff(x, i).coeff(y, j))

print("Base tensorial:")
print(Base_tensor)

print("\nPolinomio expandido en la base tensorial:")
print(prod_tensor)

print("\nMatriz de coeficientes c^{ij}:")
sp.pprint(coeff_matrix)
```

Inciso b.

```python
# Polinomios
Px = x**2 + x + 3
Py = y + 1

# Producto tensorial
prod_tensor = sp.expand(Px*Py)

# Matriz de coeficientes de los polinomios dados
coeff_matrix = sp.Matrix([
    [prod_tensor.coeff(x, i).coeff(y, j) for j in range(3)]
    for i in range(3)
])

print("\nMatriz de coeficientes c^{ij}:")
sp.pprint(coeff_matrix)
```

Inciso c.


```python
a, b = -1, 1

# Producto interno en [-1,1]
def producto(f, g):
    return sp.integrate(sp.simplify(f * g), (x, a, b))

# Ortogonalización Gram–Schmidt
def base_ort(n):
    base = [x**k for k in range(n+1)]
    ort = []
    for v in base:
        u = v
        for o in ort:
            u -= (producto(u, o) / producto(o, o)) * o
        ort.append(sp.simplify(u))
    return ort

# Obtenemos hasta grado 2
ort = base_ort(2)

print("Base Legendre hasta grado 2:")
for i, o in enumerate(ort):
    print(f"P{i}(x) = {o}")

# Calculamos coeficientes por proyección
coeficientes = []
for o in ort:
    c = producto(Px, o) / producto(o, o)
    coeficientes.append(sp.simplify(c))

# Mostramos resultados
for i, (o, c) in enumerate(zip(ort, coeficientes)):
    print(f"P{i}(x) = {o}")
    print(f"Coeficiente c{i} = {c}")
    print()

expansion = sum(c*o for c, o in zip(coeficientes, ort))

print("\nExpansión de p(x) en la base ortogonal:")
print("p(x) =", " + ".join(f"({c})*({o})" for c, o in zip(coeficientes, ort)))

print("\nVerificación (reconstrucción):", sp.simplify(expansion))
```

```python
# Polinomios de Legendre hasta grado 2
P = [sp.legendre(i, x) for i in range(3)]
Q = [sp.legendre(i, y) for i in range(3)]

print("Base de Legendre en x:", P)
print("Base de Legendre en y:", Q)

# --- Función para expandir un polinomio en la base de Legendre ---
def expand_in_legendre(poly, var, degree=2):
    """
    Expande un polinomio 'poly' en la base de Legendre hasta 'degree'.
    Devuelve lista de coeficientes [a0, a1, ..., an] tales que
    poly = sum a_i P_i(var).
    """
    coeffs = []
    for n in range(degree+1):
        Pn = sp.legendre(n, var)
        # Usamos ortogonalidad: a_n = <poly, Pn>/<Pn, Pn>
        num = sp.integrate(poly*Pn, (var, -1, 1))
        den = sp.integrate(Pn*Pn, (var, -1, 1))
        coeffs.append(sp.simplify(num/den))
    return coeffs

# --- Polinomios genéricos ---
pP = sp.Poly(sp.symbols('a0') + sp.symbols('a1')*x + sp.symbols('a2')*x**2, x).as_expr()
pG = sp.Poly(sp.symbols('b0') + sp.symbols('b1')*y + sp.symbols('b2')*y**2, y).as_expr()

print("Polinomio genérico px =", pP)
print("Polinomio genérico py =", pG)

# --- Expansión en la base de Legendre ---
a = expand_in_legendre(pP, x, 2)  # coeficientes de p^P en {P0,P1,P2}
b = expand_in_legendre(pG, y, 2)  # coeficientes de p^G en {P0,P1,P2}

print("Coeficientes de p^P en base de Legendre:", a)
print("Coeficientes de p^G en base de Legendre:", b)

# --- Construcción del tensor ---
c_tilde = sp.Matrix([[a[i]*b[j] for j in range(3)] for i in range(3)])

print("\n Matriz de coeficientes c̃^{ij}:")
print(c_tilde)

# --- Polinomio tensorial en forma expandida ---
p_tensor = sum(c_tilde[i,j]*P[i].xreplace({x:x})*Q[j].xreplace({y:y})
               for i in range(3) for j in range(3))
print("\n Polinomio tensorial expandido:")
print(sp.expand(p_tensor))

```
